### 13. Writing Files in Python
--


* `File handling` means working with files stored on the computer using Python.

* With Python we can:

  * Create a file
  * Read data from a file
  * Write data to a file
  * Append data to a file
  * Delete a file

* Files are stored on secondary storage(hard disk) and help us to save data permanently.
  * Example uses:
      * Saving user data
      * storing logs
      * writing reports
      * Saving configuration settings
    

## 2. Basic Steps for writing a file

---
When writing a file, Python follow three main steps:

1️⃣ Open the file

2️⃣ Write data into the file

3. Close the file

```python
file = open(r"C:\Users\User\Desktop\sampleData.txt","w")
file.write("Hello Siddharth")
file.close()
```

### 3. File Opening Modes
---

|Mode|Meaning|Description|
|----|----|----|
|"r"|Read|Open files for reading|
|"w"|Write|Creates file or overrites exisiting file.|
|"a"|Append|Adds data at the end of file|
|"x"|Create|Create new file, error if exists|
|"b"|Binary|Used for images/files|
|"t"|Text|Default mode|


## 4. Writing Data to a file

Mehtod-1

write() writes one string to a file

```python
file = open(r"C:\Users\User\Desktop\student.txt","w")
file.write("Name:Ram\n")
file.write("Age:25\n")
file.write("Course: Python\n")
file.close()
```

## 5. Write Multiple Lines
(Writelines)

writelines() writes multiple lines at once.

```python
file = open(r"C:\Users\User\Desktop\student.txt","w")

lines = [
    "Python is programming language\n",
    "File Handling\n",
    "Object Oriented Programming\n"
]
file.writelines(lines)
file.close()
```

### 6. Using "with" Statement

Instead of manully closing the files, Python provides with statement.

Benefits:

* Automatically close
* Safer
* Cleaner code
  

```python
with open(r"C:\Users\User\Desktop\student.txt","w") as file:
    file.write('Welcome to Python File Handling')
```

### 7. Append Data to File

--

Appends means add new data without deleting old data.

Mode Used: "a"

```python
with open(r"C:\Users\User\Desktop\student.txt","a") as file:
    file.write("\nNew User logged in")
```
