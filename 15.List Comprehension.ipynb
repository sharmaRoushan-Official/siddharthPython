### List Comprehension with if-else(conditional Expression)

Important Note:

When using if-else, the syntax changes

[express if condition true_block else false_block for item in iterable]

```python
### Even using if-else
[i if i%2==0 else 0 for i in range(1,11)]
```

Explanation:

if numbers is even --> keep it [i]

else ---> replace with 0

```python
### odd number
[i if i%2!=0 else 0 for i in range(1,11)]
```

if condition at the end ===> Filtering

if-else in the beginning = Conditional assignment

### 8. Nested List Comprehension (Multiple Loops)

syntax:


[expression for item1 in iterable1 for item2 in iterable2]

```python
paris = []
for x in [1,2,3]:
    for y in [10,20]:
        paris.append((x,y))
print(paris)
```

```python
# list Comprehension

[(x,y) for x in [1,2,3] for y in [10,20]]
```

### Nested Comprehension (2D - Matrix)

```python
[[j for j in range(3)] for i in range(4)]
```

Outer loop creates rows

Innter loop creates columns

#### List Comprehension with Functions

Exmple: Apply len() on list of words

-----

### Traditional Method


```python
words = ["python","java","c","c++","javaScript"]
print("Original Words",words)

countWords = []

for word in words:
    # print(word)
    count = 0
    for ch in word:
        count+=1
        # print(ch)
    countWords.append(count)
print("Length of Words",countWords)
```

```python
words = ["python","java","c","c++","javaScript"]
print("Original Words",words)


[len(word) for word in words]
```

#### List Comprehension with strip() and cleaning data

Example: Remove spaces from strings of list

```python
names = ["   Ravi    ","    Ajay","Kiran     ","  , ", "John            "]
print("Original String:",names)

## Traditional Method

name1 = []
for name in names:
    # print(name)
    withoutSpace = name.strip()
    name1.append(withoutSpace)
print(name1)
```

### List Comprehension

```python
names = ["   Ravi    ","    Ajay","Kiran     ","  , ", "John            "]
print("Original String:",names)


[name.strip() for name in names]
```

### List Comprehension with split()

Example: Extract words from sentence:

```python
sentence = "Python is very easy"

[word for word in sentence.split()]
```

### List Comprehension with Type Conversion

#### List Comprehension with Zip()

### combines two lists

```python
names = ["John","Alice","Anjalika"]

scores = [90,95,88]

[(n,s) for n,s in zip(names,scores)]
```

List Comprhension with flattening list



```python
matrix = [[1,2],[3,4],[5,6]]
print(matrix)

[num for row in matrix for num in row]
```

```python
*matrix
```

```python
import itertools

matrix = [[1,2],[3,4],[5,6]]
print(matrix)  # [[1, 2], [3, 4], [5, 6]]

flat = list(itertools.chain(*matrix))
print(flat)  # [1, 2, 3, 4, 5, 6]
```

```python
### 

matrix = [[1,2],[3,4],[5,6]]
print(matrix)
series = []
for row in matrix:
    for value in row:
        # print(value)
        series.append(value)
print(series)
```
