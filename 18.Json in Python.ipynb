### 5. Convert Python Dictionary to JSON  [Serialization]

json.dumps()  ---> Convert Python object to JSON String

```python
data = {
    "name":"Amit",
    "age":30,
    "city":"Hydrabad"
}

print(data)
```

```python
json_string = json.dumps(data)
json_string
```

```python
json_string["name"]
```

So, dumps() returns a JSON string()

### 6.Convert JSON to Python dictionary Deserialization

This is called Deserialization

json.loads() ---> convert JSON String to Python Obeject

```python
json_string = '{"name": "Amit", "age": 30, "city": "Hydrabad"}'

print(json_string)
```

```python
python_obj = json.loads(json_string)
print(python_obj)
```

```python
python_obj['name'] = "siddharth"
```

```python
print(python_obj)
```

```python
str_data = json.dumps(python_obj)
print(str_data)
```

### 15. JSON Encoding Problem

(Non-Seriable Objects)

Some objects cannot be converted directly to JSON

```python
from datetime import datetime
```

```python
print(datetime.now())
```

```python
datetime = {"time":datetime.now()}
print(datetime)
```

```python
print(json.dumps(datetime))
```

Solution:

Convert datetime to string

```python
from datetime import datetime
# print(datetime.now())
datetime ={"time":str(datetime.now())}
print(datetime)
print(type(datetime))
```
