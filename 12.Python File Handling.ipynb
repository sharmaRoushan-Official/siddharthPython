#### Using os module

```python
import os

```

```python
# help(os)
```

```python
if os.path.exists("C:\\Users\\User\\Desktop\\NursaryRhymes.txt"):
    print("------------File Exists----------")
    with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as file:
        print(file.read())
    
else:
    print("File Not Exists")
```

```python
# To read total number of charactars of file using python

if os.path.exists("C:\\Users\\User\\Desktop\\NursaryRhymes.txt"):
    print("------------File Exists----------")
    with open("C:\\Users\\User\\Desktop\\NursaryRhymes.txt",'r') as file:
        # print(file.read())
        content = file.read()
        countChar = 0
        spaceCount = 0
        for char in content:
            print(char)
            countChar+=1
            if char == " ":
                spaceCount  +=1
                continue
        print("Total Spaces is:",spaceCount)            
        print("Total Number of Cahracters including white space you file:",countChar)
        print(f"Total Characters without white space is:{countChar-spaceCount}")
    
else:
    print("File Not Exists")
```
