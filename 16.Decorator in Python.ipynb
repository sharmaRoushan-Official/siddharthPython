# 16. Decorators in Python

A `decorator` is a funciton that modifies or extends the behavior of another funciton(or class) without changing actual code.

Decroators are very important beacuse they are heavily used in Authentication and Authorization.

* Django/Falsk
* Logggin
* Timing functions
* validation
* Role-based access
* caching

### 1. Basic ideas of Decorator

In Python, functions are first-class objects, meaning

* You can pass a function as its argument.
* You can return a function from another function

```python
def outerFunc():
    print("Outer function running.")
    def innerFunc():
        print("Inner function running")
    innerFunc()

outerFunc()
```

```python
def greet():
    print("Hello")

def call_Function(func):
    func() # 10() # greet()

call_Function(greet)
```

### Decorator Concept

A decorator is basically:

* A funciton that takes another function
* Adds extra code 
* Return the modified function

```python
def decorator(func):
    def wrapper():
        print("Before Function Call")
        func()
        print("After Function call")
    return wrapper

def hello():
    print("Hello world")

hello = decorator(hello)
hello()
```

### 6. Decorator with @ Syntax


```python
def decorator(func):
    def wrapper():
        print("Before Function call")
        func()
        print("After funciton call")
    return wrapper
```

```python
@decorator
def hello():
    print("Hello world")
hello()
```

### Decorator with Arguments

```python
def decorator(func):
    def wrapper(name):
        print("Before Function Call")
        func(name)
        print("After Function call")
    return wrapper
```

```python
@decorator
def greet(name):
    # name = input("Enter your name: ")
    print(f"Hello {name}")
greet("Arjun")
```

### Decorator with *args[argument] and **kwargs[keyword arguments]

```python
def decorator1(func): # add
    def wrapper(*args,**kwargs):
        print("Before Function calls")
        result = func(*args, **kwargs)
        print("After function calls")
        return result
    return wrapper
```

```python
@decorator1
def add(a,b):
    return a+b
print(add(5,10))
```

```python
def decorator2(func): # add
    def wrapper(*args,**kwargs):
        print("Function is Starting")
        value = func(*args, **kwargs)
        print("Function is finished......")
        return value
    return wrapper
```

```python
@decorator2
def multiply(x,y):
    return x*y
print(multiply(5,4))
```

```python
import time
```

```python
print(time.time())
```

```python
def timer(func):
    def wrapper(*args,**kwargs):
        start = time.time()
        result = func(*args,**kwargs)
        end = time.time()
        print(f"Execution time:{end-start :.3f}")
        return result
    return wrapper
```

```python
# @timer
def task():
    time.sleep(5)
    print("Task Done")
task
```

#### Real Use of Function Decorator

```python
def logger(func):
    def wrapper(*args,**kwargs):
        print(f"Function is calling:{func.__name__}")
        return func(*args,**kwargs)
    return wrapper
```

```python
@logger
def login(user):
    print(f"{user} loggin in")

login("Ajay")
```

```python
@logger
def sumTwo(a,b):
    return a+b
sumTwo(10,15)
```

### Django Logic

```python
def login_required(func):
    def wrapper(user):
        if user == "Admin":
            return func(user)
        else:
            return "Access Denied"
    return wrapper
```

```python
@login_required
def dashboard(user):
    return f"Welcome to dashboard: {user}"
print(dashboard("Ajay"))
print(dashboard("Admin"))
```
